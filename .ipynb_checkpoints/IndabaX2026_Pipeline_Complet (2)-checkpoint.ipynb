## 🧮 2. Construction du Proxy PM2.5

### Étape 1 — ISA (Indice de Stagnation Atmosphérique)

**Référence :** Jiangsu Provincial Environmental Monitoring Center (2020) + Vovk & Kryza (2025)

$$\text{ISA} = \frac{1}{\text{wind\_max}} \times \left(1 - \frac{\min(\text{precip}, 10)}{10}\right) \times \left(1 + \frac{\max(0,\, T_{max} - 30)}{10}\right)$$

- **Terme vent** : vitesse faible → stagnation → accumulation PM2.5
- **Terme pluie** : lessivage atmosphérique (saturation à 10 mm)
- **Terme chaleur** : T° > 30°C → élévation couche limite → émission poussière

### Étape 2 — Proxy PM2.5 Composite

$$\text{PM2.5}_{proxy} = 12.0 \times \text{ISA}_{scaled} + 0.30 \times T_{mean} + 0.18 \times \text{Rad}_{scaled} \times 20 + 0.14 \times ET_0 + 7.0 \times \text{is\_harmattan} + 4.0 \times \text{is\_dry} + 3.5 \times \text{is\_stagnant} + 4.5 \times \text{is\_no\_rain}$$

````python
# ── Étape 1 : ISA ─────────────────────────────────────────────────────────────
wind  = df['wind_speed_10m_max'].clip(lower=0.5)   # éviter division par zéro
rain  = df['precipitation_sum'].clip(lower=0)
tmax  = df['temperature_2m_max']

df['isa'] = (
    (1.0 / wind)
    * (1.0 - np.minimum(rain, 10.0) / 10.0)
    * (1.0 + np.maximum(0, tmax - 30.0) / 10.0)
).clip(lower=0)

# Normalisation au percentile 95 (ISA_scaled ∈ [0, 3])
ISA_P95 = df['isa'].quantile(0.95)
df['isa_scaled'] = (df['isa'] / ISA_P95).clip(upper=3.0)

print(f'ISA — Percentile 95 (normalisation) : {ISA_P95:.4f}')
print(f'ISA — Stats : mean={df["isa"].mean():.3f} | max={df["isa"].max():.3f}')
df[['isa', 'isa_scaled']].describe().round(3)
````

````python
# ── Indicateurs discrets ──────────────────────────────────────────────────────

# Direction vent : NE = 0°-90° ou 315°-360° → Harmattan (Veselovskii 2024)
wind_dir = df['wind_direction_10m_dominant']
is_ne_wind = ((wind_dir <= 90) | (wind_dir >= 315)).astype(int)

# Région septentrionale exposée à l'Harmattan
is_north = df['region'].isin(NORTH_REGIONS).astype(int)

# Saison sèche : Novembre → Mars
is_dry = df['month'].isin([11, 12, 1, 2, 3]).astype(int)

# Harmattan actif : Nord + Sec + Vent NE
df['is_harmattan']  = (is_north & is_dry & is_ne_wind).astype(int)
df['is_dry_season'] = is_dry
df['is_stagnant']   = (df['wind_speed_10m_max'] < 3.0).astype(int)
df['is_no_rain']    = (df['precipitation_sum'] < 0.1).astype(int)

# Rayonnement normalisé (proxy AOD — Mmame & Ngongondo 2024)
rad_p95 = df['shortwave_radiation_sum'].quantile(0.95)
df['rad_scaled'] = (df['shortwave_radiation_sum'] / rad_p95).clip(upper=1.5)

print('Indicateurs construits :')
print(df[['is_harmattan','is_dry_season','is_stagnant','is_no_rain']].mean().round(3).to_string())
print(f'\n→ {df["is_harmattan"].sum():,} jours-villes avec Harmattan actif')
````

````python
# ── Étape 2 : Proxy PM2.5 composite ──────────────────────────────────────────
df['pm25_proxy'] = (
    12.0 * df['isa_scaled']                             # stagnation atmosphérique
    + 0.30 * df['temperature_2m_mean']                  # proxy hauteur couche limite
    + 0.18 * df['rad_scaled'] * 20                      # proxy AOD aérosols
    + 0.14 * df['et0_fao_evapotranspiration']           # aridité + poussière
    + 7.0  * df['is_harmattan']                         # épisode Harmattan
    + 4.0  * df['is_dry_season']                        # saison sèche générale
    + 3.5  * df['is_stagnant']                          # vent < 3 km/h
    + 4.5  * df['is_no_rain']                           # absence lessivage
).clip(lower=0)

# ── Validation physique obligatoire ──────────────────────────────────────────
north_dry  = df[df['region'].isin(NORTH_REGIONS) & (df['is_dry_season']==1)]['pm25_proxy'].mean()
south_rain = df[~df['region'].isin(NORTH_REGIONS) & (df['is_dry_season']==0)]['pm25_proxy'].mean()
ratio = north_dry / south_rain

print('=' * 55)
print(f'  Proxy PM2.5 — Statistiques globales')
print('=' * 55)
print(df['pm25_proxy'].describe().round(2).to_string())
print()
print(f'  Validation physique (Jiangsu 2020) :')
print(f'    Nord + Saison Sèche  : {north_dry:.2f}')
print(f'    Sud  + Saison Pluies : {south_rain:.2f}')
print(f'    Ratio Nord/Sud       : {ratio:.2f}  → {"✅ OK (> 1.5)" if ratio > 1.5 else "❌ KO"}')
````

````python
# ── Visualisation du Proxy PM2.5 ──────────────────────────────────────────────
fig, axes = plt.subplots(1, 3, figsize=(16, 4))

# 1 — Distribution globale
axes[0].hist(df['pm25_proxy'], bins=60, color='steelblue', edgecolor='white', alpha=0.85)
axes[0].axvline(16.55, color='#22c55e', ls='--', lw=1.5, label='Faible (16.55)')
axes[0].axvline(22.96, color='#eab308', ls='--', lw=1.5, label='Modéré (22.96)')
axes[0].axvline(30.11, color='#ef4444', ls='--', lw=1.5, label='Élevé (30.11)')
axes[0].set_xlabel('Proxy PM2.5'); axes[0].set_ylabel('Fréquence')
axes[0].set_title('Distribution globale', fontweight='bold')
axes[0].legend(fontsize=8)

# 2 — Boxplot par région
region_order = df.groupby('region')['pm25_proxy'].mean().sort_values(ascending=False).index
df.boxplot(column='pm25_proxy', by='region', ax=axes[1],
           order=region_order, vert=False, widths=0.6,
           medianprops=dict(color='red', linewidth=2))
axes[1].set_title('PM2.5 par Région', fontweight='bold')
axes[1].set_xlabel('Proxy PM2.5'); axes[1].set_ylabel('')
plt.sca(axes[1]); plt.title('')

# 3 — Évolution mensuelle
monthly_p = df.groupby('month')['pm25_proxy'].mean()
colors = ['#ef4444' if m in [11,12,1,2,3] else '#3b82f6' for m in range(1,13)]
axes[2].bar(mois, monthly_p, color=colors, edgecolor='white', alpha=0.9)
axes[2].set_ylabel('PM2.5 Proxy moyen')
axes[2].set_title('Saisonnalité du Proxy', fontweight='bold')
axes[2].legend(
    handles=[plt.Rectangle((0,0),1,1, color='#ef4444', alpha=0.8),
             plt.Rectangle((0,0),1,1, color='#3b82f6', alpha=0.8)],
    labels=['Saison sèche', 'Saison pluies'], fontsize=8
)

plt.suptitle('Proxy PM2.5 — Vue Globale', fontsize=13, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
````

## 🔧 3. Feature Engineering (49 Variables)

````python
# ── A) Encodage cyclique (Walker 2023 — continuité temporelle) ────────────────
doy = df['time'].dt.dayofyear
df['month_sin']      = np.sin(2 * np.pi * df['month'] / 12)
df['month_cos']      = np.cos(2 * np.pi * df['month'] / 12)
df['dayofyear_sin']  = np.sin(2 * np.pi * doy / 365)
df['dayofyear_cos']  = np.cos(2 * np.pi * doy / 365)
df['year']           = df['time'].dt.year

# ── B) Direction vent → vecteur unitaire (Walker 2023 — rose des vents) ───────
wr = np.deg2rad(df['wind_direction_10m_dominant'])
df['wind_dir_sin'] = np.sin(wr)
df['wind_dir_cos'] = np.cos(wr)

# ── C) Variables physiques dérivées ──────────────────────────────────────────
df['temp_amplitude']  = df['temperature_2m_max'] - df['temperature_2m_min']
df['sunshine_ratio']  = df['sunshine_duration'] / df['daylight_duration'].clip(lower=1)
df['wind_gust_ratio'] = df['wind_gusts_10m_max'] / df['wind_speed_10m_max'].clip(lower=0.1)
df['precip_log']      = np.log1p(df['precipitation_sum'])   # log(1+x) pour asymétrie

# ── D) Lags & rolling par ville (Vovk 2025 — top SHAP features) ──────────────
# CRUCIAL : groupby city pour éviter contamination entre villes
grp = df.groupby('city')
for lag in [1, 3, 7]:
    df[f'temp_lag{lag}']  = grp['temperature_2m_mean'].shift(lag)
    df[f'wind_lag{lag}']  = grp['wind_speed_10m_max'].shift(lag)
    df[f'rain_lag{lag}']  = grp['precipitation_sum'].shift(lag)
    df[f'isa_lag{lag}']   = grp['isa'].shift(lag)

df['temp_roll7']   = grp['temperature_2m_mean'].transform(lambda x: x.rolling(7, min_periods=1).mean())
df['wind_roll7']   = grp['wind_speed_10m_max'].transform(lambda x: x.rolling(7, min_periods=1).mean())
df['precip_roll7'] = grp['precipitation_sum'].transform(lambda x: x.rolling(7, min_periods=1).mean())

# ── E) Encodage géographique (tri alphabétique = stable) ─────────────────────
for col in ['city', 'region']:
    cats = sorted(df[col].unique())
    mapping = {c: i for i, c in enumerate(cats)}
    df[f'{col}_enc'] = df[col].map(mapping)

print(f'✅ Feature engineering terminé : {df.shape[1]} colonnes au total')
````

````python
# ── Liste ordonnée des features d'entraînement ────────────────────────────────
FEATURE_COLS = [
    # Météo brute
    'temperature_2m_max', 'temperature_2m_min', 'temperature_2m_mean',
    'apparent_temperature_mean',
    'precipitation_sum', 'rain_sum', 'precip_log',
    'wind_speed_10m_max', 'wind_gusts_10m_max',
    'shortwave_radiation_sum', 'et0_fao_evapotranspiration',
    'sunshine_duration', 'daylight_duration', 'precipitation_hours',
    # ISA (variable construite — Jiangsu 2020)
    'isa', 'isa_scaled',
    # Dérivées physiques
    'temp_amplitude', 'sunshine_ratio', 'wind_gust_ratio',
    'wind_dir_sin', 'wind_dir_cos',
    # Indicateurs discrets
    'is_harmattan', 'is_dry_season', 'is_stagnant', 'is_no_rain',
    # Temporel cyclique (Walker 2023)
    'month_sin', 'month_cos', 'dayofyear_sin', 'dayofyear_cos', 'year',
    # Lags 1, 3, 7 jours (Vovk 2025)
    'temp_lag1', 'temp_lag3', 'temp_lag7',
    'wind_lag1', 'wind_lag3', 'wind_lag7',
    'rain_lag1', 'rain_lag3', 'rain_lag7',
    'isa_lag1',  'isa_lag3',  'isa_lag7',
    # Rolling 7 jours
    'temp_roll7', 'wind_roll7', 'precip_roll7',
    # Géographie
    'latitude', 'longitude', 'city_enc', 'region_enc',
]
TARGET = 'pm25_proxy'

# Imputation médiane pour les NaN des lags (début de série par ville)
for col in FEATURE_COLS:
    if col in df.columns and df[col].isna().any():
        df[col] = df[col].fillna(df[col].median())

print(f'Features utilisées : {len(FEATURE_COLS)}')
print(f'Cible              : {TARGET}')
print(f'Valeurs manquantes : {df[FEATURE_COLS].isnull().sum().sum()} ✅')
````

````python
# ── Matrice de corrélation (top variables) ────────────────────────────────────
corr_cols = [
    'pm25_proxy', 'isa', 'isa_scaled', 'is_harmattan', 'is_dry_season',
    'is_no_rain', 'wind_speed_10m_max', 'precipitation_sum', 'temperature_2m_mean',
    'shortwave_radiation_sum', 'et0_fao_evapotranspiration',
]
fig, ax = plt.subplots(figsize=(10, 8))
corr = df[corr_cols].corr()
mask = np.triu(np.ones_like(corr, dtype=bool))
sns.heatmap(corr, mask=mask, annot=True, fmt='.2f', cmap='RdYlBu_r',
            center=0, ax=ax, linewidths=0.5)
ax.set_title('Matrice de Corrélation — Variables Clés', fontweight='bold', fontsize=13)
plt.tight_layout()
plt.show()
print('\n→ ISA et is_no_rain sont les meilleures corrélations avec le proxy PM2.5')
````

## ✂️ 4. Split Chronologique Train / Test

**Règle absolue (Lekinwala 2020 / Étude 5 Provinces AQHI 2023) :**
> Ne JAMAIS utiliser un split aléatoire sur des données temporelles → **data leakage** via les variables de lag.

**Split :** 80% premières observations (train) / 20% dernières (test) → ~2020-2024 train, 2025 test

````python
# ── Préparation des matrices ──────────────────────────────────────────────────
df_model = df[FEATURE_COLS + [TARGET]].dropna(subset=[TARGET]).copy()

X = df_model[FEATURE_COLS]
y = df_model[TARGET]

# Split chronologique strict (80/20)
split_idx = int(len(df_model) * 0.80)

X_train, X_test = X.iloc[:split_idx], X.iloc[split_idx:]
y_train, y_test = y.iloc[:split_idx], y.iloc[split_idx:]

print(f'Train : {len(X_train):,} obs ({len(X_train)/len(df_model)*100:.1f}%)')
print(f'Test  : {len(X_test):,} obs ({len(X_test)/len(df_model)*100:.1f}%)')
print(f'Features : {X.shape[1]}')
print()
print(f'Train — PM2.5 proxy : mean={y_train.mean():.2f} | std={y_train.std():.2f}')
print(f'Test  — PM2.5 proxy : mean={y_test.mean():.2f} | std={y_test.std():.2f}')

# Vérification : pas de fuite temporelle
train_dates = df_model['time'].iloc[:split_idx] if 'time' in df_model else None
test_dates  = df_model['time'].iloc[split_idx:] if 'time' in df_model else None
# Note : 'time' n'est pas dans FEATURE_COLS, donc df_model ne la contient pas
# La séparation est basée sur l'index (ordre chronologique assuré par sort_values)
````

## 🤖 5. Modèles Machine Learning

### Fonction d'évaluation commune

````python
# ── Dictionnaire des résultats ─────────────────────────────────────────────────
RESULTS = {}  # accumulera toutes les métriques

def evaluate_model(model, name: str, X_tr=X_train, X_te=X_test,
                   y_tr=y_train, y_te=y_test, save=True) -> dict:
    """
    Évalue un modèle et retourne ses métriques.
    Enregistre également dans RESULTS pour la comparaison finale.
    """
    y_pred = model.predict(X_te)
    r2   = r2_score(y_te, y_pred)
    mae  = mean_absolute_error(y_te, y_pred)
    rmse = float(np.sqrt(mean_squared_error(y_te, y_pred)))
    mape = float(np.mean(np.abs((y_te - y_pred) / y_te.clip(lower=1))) * 100)
    bias = float(np.mean(y_pred - y_te))

    metrics = {
        'R²'   : round(r2,   4),
        'MAE'  : round(mae,  4),
        'RMSE' : round(rmse, 4),
        'MAPE%': round(mape, 2),
        'Bias' : round(bias, 4),
    }
    if save:
        RESULTS[name] = metrics

    print(f'  [{name:<22}] R²={r2:.4f} | MAE={mae:.4f} | RMSE={rmse:.4f} | MAPE={mape:.2f}%')
    return metrics, y_pred


def plot_diagnostics(y_true, y_pred, name: str):
    """Graphiques de diagnostic : scatter réel/prédit + distribution résidus."""
    residuals = y_true.values - y_pred
    fig, axes = plt.subplots(1, 2, figsize=(13, 4))

    # Scatter réel vs prédit
    sample = np.random.choice(len(y_true), size=min(3000, len(y_true)), replace=False)
    axes[0].scatter(y_true.values[sample], y_pred[sample],
                    alpha=0.25, s=6, color='steelblue')
    lim = [min(y_true.min(), y_pred.min()), max(y_true.max(), y_pred.max())]
    axes[0].plot(lim, lim, 'r--', lw=1.5, label='Parfait')
    axes[0].set(xlabel='Réel', ylabel='Prédit', title=f'{name} — Réel vs Prédit')
    axes[0].legend(fontsize=9)

    # Distribution des résidus
    axes[1].hist(residuals, bins=60, color='salmon', edgecolor='white', alpha=0.85)
    axes[1].axvline(0, color='navy', lw=1.5, ls='--')
    axes[1].axvline(residuals.mean(), color='red', lw=1.5, ls=':')
    axes[1].set(xlabel='Résidu', title=f'{name} — Résidus (biais={residuals.mean():.3f})')

    plt.suptitle(f'Diagnostic — {name}', fontsize=12, fontweight='bold')
    plt.tight_layout()
    plt.show()


print('✅ Fonctions d\'évaluation définies')
````

### 5.1 — Lasso (Régularisation L1 + Sélection de Variables)

**Justification (Étude 5 Provinces AQHI 2023) :** Le Lasso est recommandé pour les datasets à 20+ variables colinéaires. Il contraint certains coefficients à zéro, effectuant une sélection automatique.

````python
print('Entraînement Lasso...')
t0 = datetime.now()

model_lasso = Pipeline([
    ('scaler', StandardScaler()),    # normalisation obligatoire pour Lasso
    ('model',  Lasso(
        alpha     = 0.01,            # force de régularisation L1
        max_iter  = 5000,
        random_state = RANDOM_STATE
    ))
])
model_lasso.fit(X_train, y_train)
train_time = (datetime.now() - t0).total_seconds()
print(f'  Durée : {train_time:.1f}s')

metrics_lasso, y_pred_lasso = evaluate_model(model_lasso, 'Lasso')

# Coefficients non nuls (variables sélectionnées)
coefs = model_lasso.named_steps['model'].coef_
n_nonzero = np.sum(coefs != 0)
print(f'\n  Variables sélectionnées : {n_nonzero}/{len(FEATURE_COLS)}')
top_coefs = pd.Series(np.abs(coefs), index=FEATURE_COLS).sort_values(ascending=False).head(10)
print('  Top 10 coefficients (valeur absolue) :')
print(top_coefs.round(4).to_string())

plot_diagnostics(y_test, y_pred_lasso, 'Lasso')
````

### 5.2 — ElasticNet (Régularisation L1 + L2)

````python
print('Entraînement ElasticNet...')
t0 = datetime.now()

model_en = Pipeline([
    ('scaler', StandardScaler()),
    ('model',  ElasticNet(
        alpha    = 0.01,       # régularisation globale
        l1_ratio = 0.5,        # équilibre L1/L2 (0=Ridge, 1=Lasso)
        max_iter = 5000,
        random_state = RANDOM_STATE
    ))
])
model_en.fit(X_train, y_train)
print(f'  Durée : {(datetime.now()-t0).total_seconds():.1f}s')
metrics_en, y_pred_en = evaluate_model(model_en, 'ElasticNet')
plot_diagnostics(y_test, y_pred_en, 'ElasticNet')
````

### 5.3 — Random Forest

````python
print('Entraînement Random Forest (n=200 arbres)...')
t0 = datetime.now()

model_rf = RandomForestRegressor(
    n_estimators   = 200,    # nombre d'arbres
    max_depth      = 14,     # profondeur max
    min_samples_leaf = 8,    # régularisation : feuilles minimum
    max_features   = 0.7,    # sous-échantillonnage de features
    n_jobs         = -1,     # parallélisation
    random_state   = RANDOM_STATE
)
model_rf.fit(X_train, y_train)
print(f'  Durée : {(datetime.now()-t0).total_seconds():.1f}s')
metrics_rf, y_pred_rf = evaluate_model(model_rf, 'RandomForest')

# Importance des features
importances = pd.Series(model_rf.feature_importances_, index=FEATURE_COLS).sort_values(ascending=False)

fig, axes = plt.subplots(1, 2, figsize=(15, 4))

# Top 15 features
importances.head(15).sort_values().plot(kind='barh', ax=axes[0], color='steelblue')
axes[0].set_title('Top 15 — Importance des Variables (RF)', fontweight='bold')
axes[0].set_xlabel('Importance')

# Diagnostic
sample = np.random.choice(len(y_test), 2000, replace=False)
axes[1].scatter(y_test.values[sample], y_pred_rf[sample], alpha=0.25, s=6, color='steelblue')
lim = [y_test.min(), y_test.max()]
axes[1].plot(lim, lim, 'r--', lw=1.5)
axes[1].set(xlabel='Réel', ylabel='Prédit', title='Random Forest — Réel vs Prédit')

plt.tight_layout(); plt.show()
print('\nTop 10 variables importantes :')
print(importances.head(10).round(4).to_string())
````

### 5.4 — ExtraTrees

````python
print('Entraînement ExtraTrees (n=200 arbres)...')
t0 = datetime.now()

model_et = ExtraTreesRegressor(
    n_estimators     = 200,
    max_depth        = 14,
    min_samples_leaf = 8,
    max_features     = 0.7,
    n_jobs           = -1,
    random_state     = RANDOM_STATE
)
model_et.fit(X_train, y_train)
print(f'  Durée : {(datetime.now()-t0).total_seconds():.1f}s')
metrics_et, y_pred_et = evaluate_model(model_et, 'ExtraTrees')
plot_diagnostics(y_test, y_pred_et, 'ExtraTrees')
````

### 5.5 — Gradient Boosting (GBM)

Équivalent sklearn de XGBoost — recommandé par Vovk & Kryza (2025).

````python
print('Entraînement GBM (n=150 estimateurs)...')
t0 = datetime.now()

model_gbm = GradientBoostingRegressor(
    n_estimators     = 150,    # arbres (plus lent que RF mais meilleur)
    max_depth        = 5,      # arbres peu profonds = meilleure généralisation
    learning_rate    = 0.08,   # pas d'apprentissage
    subsample        = 0.8,    # stochastic GBM
    min_samples_leaf = 8,
    random_state     = RANDOM_STATE
)
model_gbm.fit(X_train, y_train)
print(f'  Durée : {(datetime.now()-t0).total_seconds():.1f}s')
metrics_gbm, y_pred_gbm = evaluate_model(model_gbm, 'GBM')
plot_diagnostics(y_test, y_pred_gbm, 'GBM')
````

## 🧠 6. Modèles Deep Learning (MLP)

### Architecture des réseaux

```
MLP Shallow :  Input(49) → Dense(256, ReLU) → Dense(128, ReLU) → Output(1)
MLP Deep    :  Input(49) → Dense(512, ReLU) → Dense(256, ReLU) → Dense(128, ReLU) → Dense(64, ReLU) → Output(1)
```

**Paramètres :**
- Optimiseur : Adam (learning_rate adaptive)
- Régularisation : L2 (alpha=0.001) + Early Stopping (patience=10)
- Normalisation : StandardScaler obligatoire (les réseaux sont sensibles à l'échelle)

````python
print('Entraînement MLP Shallow (256 → 128)...')
t0 = datetime.now()

model_mlp1 = Pipeline([
    ('scaler', StandardScaler()),
    ('model',  MLPRegressor(
        hidden_layer_sizes  = (256, 128),   # 2 couches cachées
        activation          = 'relu',       # ReLU : meilleur que tanh pour régression
        solver              = 'adam',       # optimiseur adaptatif
        learning_rate_init  = 0.001,        # lr initiale
        max_iter            = 300,
        early_stopping      = True,         # stop si val_loss ne diminue plus
        validation_fraction = 0.1,          # 10% du train pour validation
        n_iter_no_change    = 10,           # patience = 10 epochs
        random_state        = RANDOM_STATE
    ))
])
model_mlp1.fit(X_train, y_train)
print(f'  Durée : {(datetime.now()-t0).total_seconds():.1f}s')
print(f'  Epochs effectuées : {model_mlp1.named_steps["model"].n_iter_}')

metrics_mlp1, y_pred_mlp1 = evaluate_model(model_mlp1, 'MLP_Shallow')
plot_diagnostics(y_test, y_pred_mlp1, 'MLP Shallow (256→128)')
````

````python
print('Entraînement MLP Deep (512 → 256 → 128 → 64)...')
t0 = datetime.now()

model_mlp2 = Pipeline([
    ('scaler', StandardScaler()),
    ('model',  MLPRegressor(
        hidden_layer_sizes  = (512, 256, 128, 64),  # 4 couches cachées
        activation          = 'relu',
        solver              = 'adam',
        learning_rate       = 'adaptive',           # réduction lr si plateau
        learning_rate_init  = 0.001,
        alpha               = 0.001,                # régularisation L2 (weight decay)
        max_iter            = 500,
        early_stopping      = True,
        validation_fraction = 0.1,
        n_iter_no_change    = 15,                   # patience plus longue
        random_state        = RANDOM_STATE
    ))
])
model_mlp2.fit(X_train, y_train)
print(f'  Durée : {(datetime.now()-t0).total_seconds():.1f}s')
print(f'  Epochs effectuées : {model_mlp2.named_steps["model"].n_iter_}')

metrics_mlp2, y_pred_mlp2 = evaluate_model(model_mlp2, 'MLP_Deep')
plot_diagnostics(y_test, y_pred_mlp2, 'MLP Deep (512→256→128→64)')
````

````python
# ── Prédictions superposées — tous modèles sur un échantillon ────────────────
# Sélectionner une ville et une période pour visualisation
city_viz = 'Maroua'   # ville Harmattan typique

# Indices correspondant dans le test set
# (Note : le test set est la fin du dataset, on cherche Maroua)
df_maroua = df[df['city'] == city_viz].tail(365).reset_index(drop=True)
if len(df_maroua) > 0:
    # Reconstruire les features pour cette sous-sélection
    X_maroua = df_maroua[FEATURE_COLS].copy()
    for col in FEATURE_COLS:
        if X_maroua[col].isna().any():
            X_maroua[col] = X_maroua[col].fillna(X_maroua[col].median())

    y_true_m = df_maroua['pm25_proxy'].values
    dates_m  = df_maroua['time']

    preds_m = {
        'Réel (Proxy)' : y_true_m,
        'Lasso'        : model_lasso.predict(X_maroua),
        'GBM'          : model_gbm.predict(X_maroua),
        'MLP Shallow'  : model_mlp1.predict(X_maroua),
        'MLP Deep'     : model_mlp2.predict(X_maroua),
    }

    fig, ax = plt.subplots(figsize=(14, 5))
    ax.fill_between(dates_m, y_true_m, alpha=0.15, color='steelblue')
    ax.plot(dates_m, preds_m['Réel (Proxy)'],  lw=2,   color='steelblue', label='Réel (Proxy)', zorder=5)
    ax.plot(dates_m, preds_m['Lasso'],          lw=1.2, color='#ef4444',   ls='--', label='Lasso', alpha=0.9)
    ax.plot(dates_m, preds_m['GBM'],            lw=1.2, color='#22c55e',   ls='-',  label='GBM', alpha=0.9)
    ax.plot(dates_m, preds_m['MLP Shallow'],    lw=1.2, color='#8b5cf6',   ls='-.',  label='MLP Shallow', alpha=0.9)
    ax.plot(dates_m, preds_m['MLP Deep'],       lw=1.2, color='#f97316',   ls=':',  label='MLP Deep', alpha=0.9)

    ax.axhline(30.11, ls='--', color='red', alpha=0.4, lw=1, label='Seuil élevé')
    ax.set_title(f'Prédictions Superposées — {city_viz} (dernière année)',
                 fontweight='bold', fontsize=12)
    ax.set_ylabel('PM2.5 Proxy')
    ax.legend(fontsize=8, ncol=3)
    plt.tight_layout()
    plt.show()
else:
    print(f'Ville {city_viz} non trouvée dans les données')
````
